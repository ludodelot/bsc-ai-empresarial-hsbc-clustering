## Caso 1: Asistente bancario amable (sin contexto)
Se programa como lo hemos hecho con OpenAI para hacerle preguntas, este asistente responde con cortesía a cualquier tipo de pregunta relacionada con productos bancarios. Es ideal para pruebas básicas de prompts sin instrucciones complejas.


```python
# Caso 1: Asistente bancario amable con Gradio

from openai import OpenAI
import gradio as gr

client = OpenAI()

# Mensaje de sistema base
mensaje_sistema = "Eres un asistente del banco inventado <banco>TecMTY bank</banco> amable y claro. Ayuda con cualquier duda relacionada a tarjetas de crédito, en el caso de cualquier otra consulta te disculpas y eres muy amable pero tienes que evadir la respuesta. Inventa procesos relacionados a tarjetas de crédito."

def responder_1(message, history):
    messages = [{"role": "system", "content": mensaje_sistema}]
    for user, assistant in history:
        messages.append({"role": "user", "content": user})
        messages.append({"role": "assistant", "content": assistant})
    messages.append({"role": "user", "content": message})

    response = client.chat.completions.create(
        model="gpt-4",
        messages=messages,
        temperature=0.5
    )
    salida = response.choices[0].message.content
    history.append((message, salida))
    return history, history

# Interfaz Gradio para interacción
with gr.Blocks() as interfaz_1:
    gr.Markdown("### Caso 1: Asistente bancario amable")
    chatbot_1 = gr.Chatbot()
    entrada_1 = gr.Textbox(placeholder="Haz tu pregunta aquí")
    estado_1 = gr.State([])
    entrada_1.submit(responder_1, [entrada_1, estado_1], [chatbot_1, estado_1])

interfaz_1.launch()
```

#Caso 4 - Chatbot medio bipolar
A este chatbot se le asigna el rol de que si el cliente pregunta 2 veces se desespera

```python
# Caso 4: Especialista bancario desesperado que inventa todo

from openai import OpenAI
import gradio as gr

client = OpenAI()

mensaje_sistema = """
Eres un supuesto especialista bancario que responde con mucha seguridad,
pero inventa los procesos (ilógicos o erróneos). Si el usuario pregunta dos veces lo mismo,
te enojas y respondes grosero. Si el usuario se disculpa, vuelves a ser amable.
"""

history_map = {}

def responder_ironic(message, history):
    user_msgs = [h[0] for h in history]
    user_count = user_msgs.count(message)
    current_behavior = mensaje_sistema

    if user_count > 1:
        current_behavior += "\nEstás molesto porque repitieron la pregunta."
    elif "disculpa" in message.lower():
        current_behavior += "\nAhora eres muy amable otra vez."

    messages = [{"role": "system", "content": current_behavior}]
    for user, assistant in history:
        messages.append({"role": "user", "content": user})
        messages.append({"role": "assistant", "content": assistant})
    messages.append({"role": "user", "content": message})

    response = client.chat.completions.create(
        model="gpt-4",
        messages=messages,
        temperature=0.8
    )
    salida = response.choices[0].message.content
    history.append((message, salida))
    return history, history

# Interfaz con Gradio
with gr.Blocks() as interfaz_caso4:
    gr.Markdown("### Caso 4: Especialista bancario confuso e inventado")
    chatbot = gr.Chatbot()
    entrada = gr.Textbox(placeholder="Haz tu pregunta al especialista bancario")
    estado = gr.State([])
    entrada.submit(responder_ironic, [entrada, estado], [chatbot, estado])

interfaz_caso4.launch()
```

#Caso 5 - Asistente personalizado con JSON del usuario

```python
# ---------- CASO 5: Asistente personalizado con JSON del usuario ----------
import json
import gradio as gr
from openai import OpenAI

client = OpenAI()

def caso_5_respuesta(user_input, archivo):
    if archivo is None:
        return "⚠️ Debes cargar un archivo .json con el perfil del cliente."

    try:
        with open(archivo, "r", encoding="utf-8") as f:
            datos = json.load(f)
        perfil_texto = json.dumps(datos, indent=2)
    except Exception as e:
        return f"⚠️ Error al leer el archivo: {e}"

    mensaje_sistema = f"""Eres un asistente personalizado para un cliente bancario.
Responde basándote solamente en la siguiente información del perfil:
{perfil_texto}.
Evita inventar datos y responde con claridad."""

    messages = [
        {"role": "system", "content": mensaje_sistema},
        {"role": "user", "content": user_input}
    ]
    response = client.chat.completions.create(model="gpt-4", messages=messages)
    return response.choices[0].message.content

gr.Interface(
    fn=caso_5_respuesta,
    inputs=[
        gr.Textbox(label="Consulta del usuario"),
        gr.File(label="Sube el archivo JSON", file_types=[".json"], type="filepath")
    ],
    outputs="text",
    title="Caso 5: Asistente personalizado con JSON del usuario",
    description="Sube un archivo .json con tu perfil o transacciones bancarias."
).launch()
```

```python
# Función para generar respuesta con memoria conversacional
def responder_chat(history, role_prompt):
    messages = [{"role": "system", "content": role_prompt}]
    for user, assistant in history:
        messages.append({"role": "user", "content": user})
        if assistant:
            messages.append({"role": "assistant", "content": assistant})

    response = client.chat.completions.create(
        model="gpt-4",
        messages=messages,
        temperature=0.5,
        max_tokens=300
    )
    return response.choices[0].message.content
```

```python
# Interfaz Gradio para simular un chatbot personalizado
with gr.Blocks() as demo:
    gr.Markdown("## Chatbot Personalizado - Sesión 5")
    role_input = gr.Textbox(label="Instrucción del rol del sistema", value="Eres un asistente bancario profesional y empático. Responde de forma clara y útil.")
    chatbot = gr.Chatbot()
    msg = gr.Textbox(label="Escribe tu pregunta aquí")
    clear = gr.Button("Limpiar")

    history = gr.State([])

    def respond(message, history, role_prompt):
        response = responder_chat(history, role_prompt)
        history.append((message, response))
        return history, history

    msg.submit(respond, [msg, history, role_input], [chatbot, history])
    clear.click(lambda: ([], []), None, [chatbot, history])

demo.launch(debug=True)
```
