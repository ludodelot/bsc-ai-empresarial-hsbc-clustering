## 2. Carga y preprocesamiento de datos
### 2.1 Cargar los datos desde la URL

```python
# Paso 2.1: Cargar los datos desde la URL
import pandas as pd

# Cargar el dataset desde la URL
url = "https://raw.githubusercontent.com/ArttapsTec/BasesDeDatosTEC/main/datos_default2.csv"
datos = pd.read_csv(url)

# Verificar las primeras filas del dataset
print("Primeras filas del dataset:")
print(datos.head())
```

### 2.2 Verificar y eliminar valores nulos

```python
# Paso 2.2: Verificar valores nulos
print("\nValores nulos por columna:")
print(datos.isnull().sum())

# Eliminar filas con valores nulos
datos_limpios = datos.dropna()

# Verificar que no queden valores nulos
print("\nDespués de eliminar valores nulos:")
print(datos_limpios.isnull().sum())
```

## 3. Preparación de características y variable objetivo

### 3.1 Separar las características (X) y la variable objetivo (y)

```python
# Paso 3.1: Separar características (X) y la variable objetivo (y)
X = datos_limpios.drop(columns=['BuenPago'])  # Eliminamos la columna 'BuenPago' para las características
y = datos_limpios['BuenPago']  # 'BuenPago' es nuestra variable objetivo

# Verificar las primeras filas de X (características) y y (objetivo)
print("\nPrimeras filas de X (características):")
print(X.head())
print("\nPrimeras filas de y (variable objetivo):")
print(y.head())
```

## 4. División de los datos en entrenamiento y prueba (70/30)

### 4.1 Dividir los datos usando train_test_split

```python
# Paso 4.1: Dividir los datos en entrenamiento (70%) y prueba (30%)
from sklearn.model_selection import train_test_split

# Dividir los datos en entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=18)

# Verificar tamaños de los conjuntos
print(f"\nTamaño del conjunto de entrenamiento: {X_train.shape}")
print(f"Tamaño del conjunto de prueba: {X_test.shape}")
```

# 5. Creación de modelos base de Random Forest

### 5.1 Crear los modelos de Random Forest con diferentes profundidades

```python
# Paso 5.1: Crear los modelos de Random Forest
from sklearn.ensemble import RandomForestClassifier

# Definir los modelos con diferentes profundidades (max_depth)
rf1 = RandomForestClassifier(max_depth=2, n_estimators=20, random_state=18)
rf2 = RandomForestClassifier(max_depth=3, n_estimators=20, random_state=18)
rf3 = RandomForestClassifier(max_depth=4, n_estimators=20, random_state=18)
rf4 = RandomForestClassifier(max_depth=5, n_estimators=20, random_state=18)
rf5 = RandomForestClassifier(max_depth=6, n_estimators=20, random_state=18)
rf6 = RandomForestClassifier(max_depth=7, n_estimators=20, random_state=18)

# Guardar los modelos en una lista
modelos = [rf1, rf2, rf3, rf4, rf5, rf6]
```

# 6. Definición de la métrica personalizada

### 6.1 Crear la función para la métrica personalizada

```python
# Paso 6.1: Crear la métrica personalizada para evaluar los modelos
from sklearn.metrics import confusion_matrix

def metrica_personalizada(y_true, y_pred, peso_tp=2, peso_tn=4, peso_fp=-1, peso_fn=3):
    """
    Calcula una métrica personalizada basada en la matriz de confusión:
    - Verdaderos Positivos (TP), Verdaderos Negativos (TN), Falsos Positivos (FP), Falsos Negativos (FN).
    """
    tn, fp, fn, tp = confusion_matrix(y_true, y_pred).ravel()
    
    # Cálculo de la métrica personalizada
    score = tp * peso_tp + tn * peso_tn + fp * peso_fp + fn * peso_fn
    return score
```

# 7. Entrenamiento y evaluación de los modelos

```python
# Paso 7.1: Entrenar y evaluar todos los modelos, y preparar los datos para las gráficas
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import accuracy_score, recall_score, precision_score, roc_auc_score

# Lista para almacenar los resultados de todos los modelos
resultados_modelos = []

# Evaluar todos los modelos con un bucle
for i, modelo in enumerate(modelos, 1):
    print(f"\nEntrenando y evaluando el Modelo {i} (max_depth={modelo.max_depth})")
    
    # Entrenar el modelo con los datos de entrenamiento
    modelo.fit(X_train, y_train)
    
    # Realizar predicciones sobre el conjunto de prueba
    y_pred = modelo.predict(X_test)
    y_prob = modelo.predict_proba(X_test)[:, 1]  # Probabilidades para la clase 1
    
    # Calcular las métricas de desempeño
    accuracy = accuracy_score(y_test, y_pred)  # Precisión
    recall = recall_score(y_test, y_pred)  # Recall
    precision = precision_score(y_test, y_pred)  # Precisión
    roc_auc = roc_auc_score(y_test, y_prob)  # ROC-AUC
    
    # Calcular la métrica personalizada
    score_personalizado = metrica_personalizada(y_test, y_pred)
    
    # Almacenar los resultados del modelo en un diccionario
    resultados_modelos.append({
        'Modelo': f"RandomForest (max_depth={modelo.max_depth})",
        'Precisión': accuracy,
        'Recall': recall,
        'Precision': precision,
        'ROC-AUC': roc_auc,
        'Métrica Personalizada': score_personalizado
    })
    
    # Imprimir los resultados para cada modelo
    print(f"Modelo {i} - Precisión: {accuracy:.4f}")
    print(f"Modelo {i} - Recall: {recall:.4f}")
    print(f"Modelo {i} - Precision: {precision:.4f}")
    print(f"Modelo {i} - ROC-AUC: {roc_auc:.4f}")
    print(f"Modelo {i} - Métrica Personalizada: {score_personalizado:.4f}")
```

# 8. Almacenamiento y Visualización de Resultados

### 8.1 Guardar los resultados de los modelos en un DataFrame

```python
# Paso 8.1: Guardar los resultados de todos los modelos en un DataFrame
import pandas as pd

# Convertir los resultados a un DataFrame
resultados_df = pd.DataFrame(resultados_modelos)

# Mostrar los resultados en una tabla
resultados_df
# También imprimimos los resultados en consola para revisión
print("\nResultados de todos los modelos:")
print(resultados_df)
```

# 9. Graficar las Métricas Comparativas

### 9.1 Graficar las métricas de los modelos

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set(style="whitegrid", palette="muted")
plt.figure(figsize=(16, 12))

# Métricas que deseas graficar
metricas = [
    ('Precisión', 'Comparación de Precisión'),
    ('Recall', 'Comparación de Recall'),
    ('Precision', 'Comparación de Precision'),
    ('ROC-AUC', 'Comparación de ROC-AUC'),
    ('Métrica Personalizada', 'Comparación de Métrica Personalizada')
]

# Crear una gráfica por cada métrica
for i, (columna, titulo) in enumerate(metricas, start=1):
    plt.subplot(2, 3, i)
    ax = sns.barplot(x='Modelo', y=columna, data=resultados_df, palette='coolwarm')
    plt.title(titulo, fontsize=14)
    plt.xlabel('Modelo', fontsize=12)
    plt.ylabel(columna, fontsize=12)
    plt.xticks(rotation=45, ha='right')
    for p in ax.patches:
        ax.annotate(f'{p.get_height():.2f}', (p.get_x() + p.get_width() / 2., p.get_height()),
                    ha='center', va='center', fontsize=10, color='black',
                    xytext=(0, 5), textcoords='offset points')

plt.tight_layout()
plt.show()
```
